# Imported the python libraries.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# Took the data from github and converted it into dataframes. Used head to show first five column


```python
url = "https://raw.githubusercontent.com/vincentarelbundock/Rdatasets/master/csv/AER/TeachingRatings.csv"

df = pd.read_csv(url)

df.head()
```

# Calculated the no. of rows and columns in the dataset.And listed the names of the columns.

```python
print("Rows and Columns:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
```

# The dataset is cross-sectional as the taken dataset is from one point of time and do not varies by time

## Calculated the mean, median, minimum and maximum

```python
print("Mean:", df["students"].mean())
print("Median:", df["students"].median())
print("Minimum:", df["students"].min())
print("Maximum:", df["students"].max())
```

# Performed the descriptive statistics using describe.

```python
df.describe()
```

# Made a histogram using beauty and frequency columns and made 20 bins.

```python
plt.hist(df["beauty"], bins=20, edgecolor="black")

plt.xlabel("Beauty")
plt.ylabel("Frequency")
plt.title("Distribution of Beauty Scores")

plt.show()
```

# Grouped the gender and beauty columns and calculated mean and standard deviation.

```python
gender_stats = df.groupby("gender")["beauty"].agg(["mean", "std"])

gender_stats
```

# Rounded of the values to 2 decimal points.

```python
gender_stats.round(2)
```

#Took tenure and gender column and found the percentage of the proportions.


```python
tenure_table = pd.crosstab(
    df["gender"],
    df["tenure"],
    normalize="index"
) * 100

tenure_table.round(2)
```

# The percentage of tenured professors is more as compared to female.
