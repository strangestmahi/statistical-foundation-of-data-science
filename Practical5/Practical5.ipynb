# This cell imports pandas and loads the TeachingRatings dataset.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/vincentarelbundock/Rdatasets/master/csv/AER/TeachingRatings.csv"
df = pd.read_csv(url)

df.head()
```

# This calculates the proportion of teachers having an evaluation score greater than 4.5.

```python
probability = (df['eval'] > 4.5).mean()

print("Probability =", probability)
```

# This finds the probability of evaluation scores lying between 3.5 and 4.2.

```python
probability = ((df['eval'] > 3.5) & (df['eval'] < 4.2)).mean()

print("Probability =", probability)
```

# This cell stores the given population mean, standard deviation, sample size, and sample mean.

```python
mu = 12
sigma = 5.5
n = 36
x_bar = 10.7
```

# This calculates the Z-test statistic using the given sample information.

```python
import math

z = (x_bar - mu) / (sigma / math.sqrt(n))

print("Z =", z)
```

# For a two-tailed test at 5% significance level, the critical values are ±1.96.

```python
critical_value = 1.96

print("Critical values =", -critical_value, "and", critical_value)
```

# For a two-tailed test at 5% significance level, the critical values are ±1.96.

```python
if abs(z) > critical_value:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

# The calculated Z-value is compared with the critical value to make the final decision.
